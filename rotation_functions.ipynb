```python
import numpy as np

# Define generic functions for rotating vectors about an axis
def rotate_vector(v, axis, theta_deg):
    theta = np.radians(theta_deg)
    c = np.cos(theta)
    s = np.sin(theta)
    t = 1 - c

    R = np.array([
        [t*axis[0]**2 + c             , t*axis[0]*axis[1] - s*axis[2], t*axis[0]*axis[2] + s*axis[1]],
        [t*axis[0]*axis[1] + s*axis[2], t*axis[1]**2 + c             , t*axis[1]*axis[2] - s*axis[0]],
        [t*axis[0]*axis[2] - s*axis[1], t*axis[1]*axis[2] + s*axis[0], t*axis[2]**2 + c]
    ])
    return np.dot(R, v)

def rotate_two_vectors(v_point_0,theta_1,theta_2,axis_1,axis_2):
    # rotates pointing about the first rotation axis
    v_point_1 = rotate_vector(v_point_0,axis_1,theta_1)

    # rotates second rotation axis about the first rotation axis
    axis_2_rotated = rotate_vector(axis_2,axis_1,theta_1)
    
    # rotates pointing about the rotated second rotation axis
    v_point_2 = rotate_vector(v_point_1,axis_2_rotated,theta_2)
    return v_point_2

def angle_between_vectors(v1, v2):
    # Compute the dot product and magnitudes
    dot_product = np.dot(v1, v2)
    magnitude_v1 = 1
    magnitude_v2 = 1
    
    # Calculate the cosine of the angle and clip to avoid precision errors
    cosine_angle = dot_product / (magnitude_v1 * magnitude_v2)
    cosine_angle = np.clip(cosine_angle, -1.0, 1.0)
    
    # Take arccos and convert from radians to degrees
    angle_radians = np.arccos(cosine_angle)
    angle_degrees = np.degrees(angle_radians)
    
    return angle_degrees
```
